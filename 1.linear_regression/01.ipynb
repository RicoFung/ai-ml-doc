# 线性回归

这个 notebook 展示最基础的一元线性回归：使用特征 $x$ 预测标签 $y$。

## 1. 数据准备

设特征为房屋面积 $x$（单位：平方米），标签为房价 $y$（单位：万元）。下面随机生成 10 行更贴近现实的样本数据：

| 编号 | 面积 $x$（平方米） | 房价 $y$（万元） |
| --- | --- | --- |
| 1 | 58 | 132 |
| 2 | 62 | 145 |
| 3 | 67 | 156 |
| 4 | 71 | 167 |
| 5 | 76 | 181 |
| 6 | 83 | 198 |
| 7 | 88 | 210 |
| 8 | 94 | 226 |
| 9 | 101 | 245 |
| 10 | 108 | 261 |


## 2. 线性回归公式

线性回归模型为：

$$
\hat{y} = wx + b
$$

其中：

- $w$ 是权重（斜率）
- $b$ 是偏置（截距）
- $\hat{y}$ 是模型预测值


## 3. 代价函数（Cost Function）

在线性回归中，常用**均方误差** (**MSE**) 形式的代价函数：

$$
J(w, b) = \frac{1}{2m} \sum_{i=1}^{m} (\hat{y}^{(i)} - y^{(i)})^2
$$

其中：

- $m$ 表示样本数量
- $\hat{y}^{(i)}$ 表示第 $i$ 个样本的预测值
- $y^{(i)}$ 表示第 $i$ 个样本的真实值
- $\sum$ 表示求和符号，读作“sigma”，例如：$$ \sum_{i=1}^{3} y^{(i)} = y^{(1)} + y^{(2)} + y^{(3)} = 132 + 145 + 156 $$


## 4. 更新 $w$ 和 $b$ 的公式

使用梯度下降时，参数更新公式为：

$$
w := w - \alpha \frac{1}{m} \sum_{i=1}^{m} (\hat{y}^{(i)} - y^{(i)}) x^{(i)}
$$

$$
b := b - \alpha \frac{1}{m} \sum_{i=1}^{m} (\hat{y}^{(i)} - y^{(i)})
$$

其中 $\alpha$ 是学习率。

## 5. $w$ 和 $b$ 的训练过程示例

下面用这 10 组房屋面积与房价样本说明一次梯度下降中，$w$ 和 $b$ 是如何更新的。为了便于展示，先假设参数初始值为：

$$
w = 0, \quad b = 0
$$

同时，给学习率一个具体的值。这里先设：

$$
\alpha = 0.0001
$$

学习率表示每次参数更新时前进的步长。由于这里的面积和房价数值都比前面的玩具示例更大，所以把 $\alpha$ 设为更小的 $0.0001$，避免更新步长过大。

此时模型预测公式为：

$$
\hat{y} = wx + b = 0 \cdot x + 0 = 0
$$

也就是说，当前对所有样本的预测值都是 0。

***第一步：代入代价函数***

共有 $m = 10$ 个样本，因此代价函数为：

$$
J(w, b) = \frac{1}{2 \times 10} \sum_{i=1}^{10} (\hat{y}^{(i)} - y^{(i)})^2
$$

当 $w=0, b=0$ 时，所有 $\hat{y}^{(i)} = 0$，代入数据得到：

$$
J(0,0) = \frac{1}{20} \left[(0-132)^2 + (0-145)^2 + (0-156)^2 + (0-167)^2 + (0-181)^2 + (0-198)^2 + (0-210)^2 + (0-226)^2 + (0-245)^2 + (0-261)^2\right]
$$

$$
= \frac{385961}{20} = 19298.05
$$

***第二步：计算 $w$ 的梯度***

$$
\frac{1}{m} \sum_{i=1}^{m} (\hat{y}^{(i)} - y^{(i)}) x^{(i)}
$$

代入这 10 个样本，且当前 $\hat{y}^{(i)} = 0$：

$$
\frac{1}{10} \left[(0-132) \cdot 58 + (0-145) \cdot 62 + (0-156) \cdot 67 + (0-167) \cdot 71 + (0-181) \cdot 76 + (0-198) \cdot 83 + (0-210) \cdot 88 + (0-226) \cdot 94 + (0-245) \cdot 101 + (0-261) \cdot 108\right]
$$

进一步写成：

$$
\frac{1}{10} \left[-7656 -8990 -10452 -11857 -13756 -16434 -18480 -21244 -24745 -28188\right]
$$

$$
= \frac{-161802}{10} = -16180.2
$$

因此，$w$ 的更新公式变成：

$$
w := w - \alpha(-16180.2) = w + 16180.2\alpha
$$

把 $\alpha = 0.0001$ 代入：

$$
w := 0 + 0.0001 \times 16180.2 = 1.61802
$$

所以这一次更新后：

$$
w = 1.61802
$$

***第三步：计算 $b$ 的梯度***

$$
\frac{1}{m} \sum_{i=1}^{m} (\hat{y}^{(i)} - y^{(i)})
$$

代入这 10 个样本，且当前 $\hat{y}^{(i)} = 0$：

$$
\frac{1}{10} \left[(0-132) + (0-145) + (0-156) + (0-167) + (0-181) + (0-198) + (0-210) + (0-226) + (0-245) + (0-261)\right]
$$

进一步写成：

$$
\frac{1}{10} \left[-132 -145 -156 -167 -181 -198 -210 -226 -245 -261\right]
$$

$$
= \frac{-1921}{10} = -192.1
$$

因此，$b$ 的更新公式变成：

$$
b := b - \alpha(-192.1) = b + 192.1\alpha
$$

把 $\alpha = 0.0001$ 代入：

$$
b := 0 + 0.0001 \times 192.1 = 0.01921
$$

所以这一次更新后：

$$
b = 0.01921
$$

***第四步：用表格理解训练过程***

| 阶段 | $w$ | $b$ | 代价函数 $J(w,b)$ | 说明 |
| --- | --- | --- | --- | --- |
| 初始状态 | 0 | 0 | 19298.05 | 所有样本的预测值都为 0，误差很大。 |
| 第 1 次更新后 | 1.61802 | 0.01921 | $\approx 1998.25$ | 做完一次梯度下降后，参数朝着更优方向移动，代价函数明显下降。 |
| 继续迭代后 | 逐步接近 2.57 | 逐步接近 -15.56 | 继续下降 | 每一轮都用新的 $w$ 和 $b$ 重新计算预测值、代价函数和梯度，再继续更新参数。 |
| 最终收敛时 | $\approx 2.570106$ | $\approx -15.562576$ | $\approx 0.404175$ | 模型已经找到一条很适合这组样本的直线，误差降到很小。 |

从这个表格可以更直观地看到：梯度下降的核心，就是让参数 $w$ 和 $b$ 持续更新，使代价函数 $J(w,b)$ 一步一步变小。

## 6. 如何判断收敛

在线性回归训练中，常见的一种判断方法是：看这一轮和上一轮的代价函数差值是否已经足够小。

判断公式为：

$$
\left| J_{\text{prev}} - J(w,b) \right| < \epsilon
$$

其中：

- $J_{\text{prev}}$ 表示上一轮的代价函数值
- $J(w,b)$ 表示当前这一轮的代价函数值
- $\epsilon$ 表示收敛阈值，是一个很小的正数

这条规则的意思是：如果这一轮的代价值和上一轮相比，已经几乎没有明显变化了，就可以认为模型基本学得差不多了，可以停止训练。

例如，当我们设定：

$$
\epsilon = 0.0001
$$

如果出现：

$$
\left| J_{\text{prev}} - J(w,b) \right| = 0.00003 < 0.0001
$$

那么就可以判断：模型已经收敛，可以停止迭代。

## 7. 训练过程总结

线性回归的训练目标，就是不断调整参数 $w$ 和 $b$，让代价函数 $J(w, b)$ 尽可能小。

| 符号 | 释义 |
| --- | --- |
| $$\{(x^{(1)}, y^{(1)}), (x^{(2)}, y^{(2)}), \ldots, (x^{(m)}, y^{(m)})\}$$ | 训练数据 |
| $x$ | 特征值 |
| $y$ | 真实值 |
| $\hat{y}$ | 预测值 |
| $w$ | 权重 |
| $b$ | 偏置 |
| $\frac{\partial J(w,b)}{\partial w}$ | 对 $w$ 的梯度 |
| $\frac{\partial J(w,b)}{\partial b}$ | 对 $b$ 的梯度 |
| $\alpha$ | 学习率 |
| $\text{num\_iters}$ | 最大迭代次数 |
| $\epsilon$ | 收敛阈值 |
| $J(w,b)$ | 代价函数 |
| $J_{\text{prev}}$ | 上一轮的代价函数值 |

$J_{\text{prev}}$ 一开始设为正无穷，是因为第一轮训练时还没有“上一轮”的代价值可以比较，所以先给它一个特别大的初始值。这样第一次判断收敛时，就不会误以为模型已经收敛。

在 Python 代码里，正无穷通常写成：`float('inf')`。

***初始化：***

$$
w = 0, \quad b = 0, \quad J_{\text{prev}} = +\infty
$$

***for (i=0, i<num_iters, i++):***

***{***

$$
\begin{aligned}
&\text{1. 计算当前“代价函数”（其中 } \hat{y}^{(i)} = wx^{(i)} + b {）: } \\
&J(w, b) = \frac{1}{2m} \sum_{i=1}^{m} (\hat{y}^{(i)} - y^{(i)})^2 = \frac{1}{2m} \sum_{i=1}^{m} (wx^{(i)} + b - y^{(i)})^2\\[6pt]
&\text{2. 判断“收敛”：}\\
&\text{若 } \left| J_{\text{prev}} - J(w,b) \right| < \epsilon \text{，则停止训练；否则令 } J_{\text{prev}} = J(w,b)\\[6pt]
&\text{3. 若未收敛，则更新“参数”（w 和 b）：}\\
&w = w - \alpha \frac{\partial J(w,b)}{\partial w} = w - \alpha \frac{1}{m} \sum_{i=1}^{m} (\hat{y}^{(i)} - y^{(i)})x^{(i)} = w - \alpha \frac{1}{m} \sum_{i=1}^{m} (wx^{(i)} + b - y^{(i)})x^{(i)}\\[6pt]
&b = b - \alpha \frac{\partial J(w,b)}{\partial b} = b - \alpha \frac{1}{m} \sum_{i=1}^{m} (\hat{y}^{(i)} - y^{(i)}) = b - \alpha \frac{1}{m} \sum_{i=1}^{m} (wx^{(i)} + b - y^{(i)})\\
\end{aligned}
$$

***}***

最终输出训练后的 $w$ 和 $b$。

## 8. 示例代码

下面给出一个完整的线性回归训练示例，包含代价函数、梯度下降、模型训练和预测过程。

```python
# 代价函数：计算当前参数下的平均平方误差的一半
def cost_function(x, y, w, b):
    m = len(x)
    total_cost = 0.0

    for i in range(m):
        # 根据当前的 w 和 b 计算第 i 个样本的预测值
        y_hat = w * x[i] + b
        total_cost += (y_hat - y[i]) ** 2

    return total_cost / (2 * m)
```

```python
# 梯度下降：根据所有样本的梯度更新一次参数
def gradient_descent(x, y, w, b, alpha):
    m = len(x)
    dj_dw = 0.0
    dj_db = 0.0

    for i in range(m):
        # 先计算预测误差，再分别累加到 w 和 b 的梯度上
        y_hat = w * x[i] + b
        error = y_hat - y[i]
        dj_dw += error * x[i]
        dj_db += error

    dj_dw = dj_dw / m
    dj_db = dj_db / m

    w = w - alpha * dj_dw
    b = b - alpha * dj_db

    return w, b
```

```python
# 模型训练：先计算代价函数，再判断收敛，未收敛时再更新参数
def train_model(x, y, alpha=0.00025, num_iters=500000, epsilon=1e-9):
    w = 0.0
    b = 0.0
    j_prev = float('inf')
    history = []
    last_iter = 0
    last_cost_diff = float('inf')
    prev_log = None
    current_log = None

    def log_iteration(iter_idx, w, b, current_cost, cost_diff):
        # 输出当前迭代的参数、代价函数和收敛阈值信息
        print(
            f"[{iter_idx:>4}/{num_iters}]: w = {w:.6f}, b = {b:.6f}, cost = {current_cost:.12f}, "
            f"j_prev - current_cost = {cost_diff}, epsilon = {epsilon}"
        )

    for i in range(num_iters):
        # 1. 计算当前代价函数值
        current_cost = cost_function(x, y, w, b)
        history.append(current_cost)

        cost_diff = j_prev - current_cost if j_prev != float('inf') else float('inf')
        last_iter = i
        last_cost_diff = cost_diff
        current_log = (i, w, b, current_cost, cost_diff)

        if i % 50000 == 0:
            log_iteration(*current_log)

        # 2. 判断是否收敛
        if abs(j_prev - current_cost) < epsilon:
            if prev_log is not None and prev_log[0] == i - 1 and (i - 1) % 5000 != 0:
                log_iteration(*prev_log)
            break

        j_prev = current_cost

        # 3. 若未收敛，则使用梯度下降更新参数
        w, b = gradient_descent(x, y, w, b, alpha)
        prev_log = current_log

    log_iteration(last_iter, w, b, history[-1], last_cost_diff)

    return w, b, history
```

```python
# 使用前面定义的函数训练模型并预测房价
x = [58, 62, 67, 71, 76, 83, 88, 94, 101, 108]
y = [132, 145, 156, 167, 181, 198, 210, 226, 245, 261]

w, b, history = train_model(x, y, alpha=0.00025, num_iters=1000000, epsilon=1e-9)

print(f"【训练完成】: w = {w:.6f}, b = {b:.6f}, cost = {history[-1]:.12f}")

# 使用训练后的模型预测不同面积房屋的价格
x_test = [58, 62, 120]
predictions = [w * value + b for value in x_test]

for value, pred in zip(x_test, predictions):
    print(f"面积 = {value} 平方米, 【预测房价】 = {pred:.2f} 万元")
```
