# 特征工程和多项式回归

这一节用房价数据说明：除了直接使用原始特征，还可以通过特征工程构造新特征，让线性回归表达更复杂的关系。

## 1. 什么是特征工程

特征工程就是根据已有数据构造更有用的新特征。例如，原始数据中有面积、卧室数、楼层、房龄，我们可以构造：

| 新特征 | 含义 |
| --- | --- |
| 面积平方 | 表示房价和面积可能不是简单直线关系 |
| 房龄平方 | 表示房龄变大后，影响可能加速变化 |
| 面积 / 卧室数 | 表示平均每个卧室占用面积 |

多项式回归就是一种常见的特征工程：把 $x$、$x^2$、$x^3$ 等作为新的输入特征。

```python
import numpy as np
import matplotlib.pyplot as plt


data = np.genfromtxt("data/data_house.csv", delimiter=",", skip_header=1)

X = data[:, :4]
y = data[:, 4]

area = X[:, 0]

print(f"样本数量: {X.shape[0]}")
print(f"首行原始特征: {X[0]}")
print(f"首行房价: {y[0]} 万元")
```

## 2. 多项式特征

为了让例子更直观，这里先只使用面积 $x$ 预测房价 $y$。

普通线性回归使用：

$$
f(x)=w_1x+b
$$

二次多项式回归使用：

$$
f(x)=w_1x+w_2x^2+b
$$

它本质上仍然是线性回归，只是输入特征从 $[x]$ 变成了 $[x, x^2]$。

```python
X_area = area.reshape(-1, 1)
X_area_poly = np.column_stack([area, area ** 2])

print("普通线性回归前 3 行特征：")
print(X_area[:3])

print("\n二次多项式回归前 3 行特征：")
print(X_area_poly[:3])
```

## 3. 对比普通线性回归和多项式回归

多项式特征的数值可能变得很大，例如 $area^2$，所以训练前仍然需要做特征缩放。

```python
def zscore_normalize_features(X):
    mu = np.mean(X, axis=0)
    sigma = np.std(X, axis=0)
    X_scaled = (X - mu) / sigma
    return X_scaled, mu, sigma


def compute_cost(X, y, w, b):
    m = len(y)
    error = X @ w + b - y
    return np.sum(error ** 2) / (2 * m)


def gradient_descent(X, y, alpha, num_iters):
    m, n = X.shape
    w = np.zeros(n)
    b = 0.0
    cost_history = []

    for _ in range(num_iters):
        error = X @ w + b - y
        dj_dw = X.T @ error / m
        dj_db = np.sum(error) / m

        w = w - alpha * dj_dw
        b = b - alpha * dj_db
        cost_history.append(compute_cost(X, y, w, b))

    return w, b, cost_history
```

```python
X_area_scaled, area_mu, area_sigma = zscore_normalize_features(X_area)
X_area_poly_scaled, poly_mu, poly_sigma = zscore_normalize_features(X_area_poly)

alpha = 0.1
num_iters = 300

w_linear, b_linear, cost_linear = gradient_descent(X_area_scaled, y, alpha, num_iters)
w_poly, b_poly, cost_poly = gradient_descent(X_area_poly_scaled, y, alpha, num_iters)

print(f"普通线性回归最终代价: {cost_linear[-1]:.2f}")
print(f"二次多项式回归最终代价: {cost_poly[-1]:.2f}")
```

```python
area_line = np.linspace(area.min(), area.max(), 200)

area_line_linear = area_line.reshape(-1, 1)
area_line_linear_scaled = (area_line_linear - area_mu) / area_sigma
prediction_linear = area_line_linear_scaled @ w_linear + b_linear

area_line_poly = np.column_stack([area_line, area_line ** 2])
area_line_poly_scaled = (area_line_poly - poly_mu) / poly_sigma
prediction_poly = area_line_poly_scaled @ w_poly + b_poly

plt.figure(figsize=(8, 4))
plt.scatter(area, y, s=12, alpha=0.55, label="data")
plt.plot(area_line, prediction_linear, label="linear")
plt.plot(area_line, prediction_poly, label="polynomial degree 2")
plt.xlabel("area")
plt.ylabel("price")
plt.title("Linear vs polynomial regression")
plt.legend()
plt.grid(True, alpha=0.25)
plt.show()
```

## 4. 多特征中的特征工程

在多元线性回归中，也可以构造更多特征。例如：

$$
[x_1, x_2, x_3, x_4]
$$

可以扩展为：

$$
[x_1, x_2, x_3, x_4, x_1^2, x_4^2, \frac{x_1}{x_2}]
$$

这样模型仍然是线性回归，但它能表达更复杂的关系。

```python
area = X[:, 0]
bedrooms = X[:, 1]
age = X[:, 3]

area_per_bedroom = area / bedrooms
X_engineered = np.column_stack([X, area ** 2, age ** 2, area_per_bedroom])

print(f"原始特征数量: {X.shape[1]}")
print(f"特征工程后特征数量: {X_engineered.shape[1]}")
print(f"首行扩展特征: {X_engineered[0]}")
```

## 5. 小结

| 概念 | 作用 |
| --- | --- |
| 特征工程 | 根据已有特征构造更有表达力的新特征 |
| 多项式回归 | 把 $x^2$、$x^3$ 等多项式项作为新特征 |
| 特征缩放 | 防止新特征数值过大，让梯度下降更稳定 |

多项式回归并没有改变线性回归的训练方式，它改变的是输入给模型的特征。
