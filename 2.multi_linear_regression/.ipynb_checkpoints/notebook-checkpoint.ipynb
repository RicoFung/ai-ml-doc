# 多元线性回归

这个 notebook 展示最基础的多元线性回归：使用多个房屋特征 $x_1, x_2, x_3, x_4$ 来预测标签 $y$。本案例中，$x_1$ 表示面积，$x_2$ 表示卧室数，$x_3$ 表示楼层，$x_4$ 表示房龄，$y$ 表示房价。

## 1. 数据准备

设每个房屋样本有四个特征：$x_1$（面积，平方米）、$x_2$（卧室数）、$x_3$（楼层）、$x_4$（房龄，年），标签 $y$ 表示房价（万元）。下面给出 10 行样本数据：

| 编号 | $x_1$ 面积 | $x_2$ 卧室 | $x_3$ 楼层 | $x_4$ 房龄 | $y$ 房价 |
| --- | --- | --- | --- | --- | --- |
| 1 | 60 | 2 | 5 | 10 | 167.5 |
| 2 | 75 | 3 | 8 | 5 | 227 |
| 3 | 90 | 3 | 12 | 3 | 267 |
| 4 | 110 | 4 | 15 | 1 | 330.5 |
| 5 | 55 | 2 | 3 | 20 | 134.5 |
| 6 | 130 | 4 | 20 | 2 | 376 |
| 7 | 85 | 3 | 6 | 12 | 230 |
| 8 | 100 | 3 | 10 | 8 | 274 |
| 9 | 70 | 2 | 18 | 0 | 227 |
| 10 | 120 | 4 | 25 | 6 | 355.5 |

这组数据可以理解为满足下面这条线性关系：

$$
y = 2x_1 + 15x_2 + 1.5x_3 - 2x_4 + 30
$$

其中，面积、卧室数和楼层对房价是正向影响，房龄对房价是负向影响。

```python
X = [
    [60, 2, 5, 10],
    [75, 3, 8, 5],
    [90, 3, 12, 3],
    [110, 4, 15, 1],
    [55, 2, 3, 20],
    [130, 4, 20, 2],
    [85, 3, 6, 12],
    [100, 3, 10, 8],
    [70, 2, 18, 0],
    [120, 4, 25, 6],
]

y = [167.5, 227, 267, 330.5, 134.5, 376, 230, 274, 227, 355.5]

list(zip(X, y))
```

## 2. 示例代码

下面给出一个完整的多特征房价预测训练示例，包含代价函数、梯度下降、模型训练和预测过程。由于面积的数值量级比卧室数、楼层、房龄更大，学习率需要取得更小一些，避免参数更新过大。

```python
# 代价函数：计算当前参数下的平均平方误差的一半
def cost_function(X, y, w, b):
    m = len(X)
    total_cost = 0.0

    for i in range(m):
        # 计算第 i 个样本的预测值
        y_hat = sum(w[j] * X[i][j] for j in range(len(w))) + b
        total_cost += (y_hat - y[i]) ** 2

    return total_cost / (2 * m)
```

```python
# 梯度下降：根据所有样本的梯度更新一次参数
def gradient_descent(X, y, w, b, alpha):
    m = len(X)
    n = len(w)
    dj_dw = [0.0] * n
    dj_db = 0.0

    for i in range(m):
        # 先计算预测误差，再分别累加到每个参数的梯度上
        y_hat = sum(w[j] * X[i][j] for j in range(n)) + b
        error = y_hat - y[i]

        for j in range(n):
            dj_dw[j] += error * X[i][j]

        dj_db += error

    dj_dw = [grad / m for grad in dj_dw]
    dj_db = dj_db / m

    for j in range(n):
        w[j] = w[j] - alpha * dj_dw[j]

    b = b - alpha * dj_db

    return w, b
```

```python
# 模型训练：先计算代价函数，再判断收敛，未收敛时再更新参数
def train_model(X, y, alpha=1e-6, num_iters=100000, epsilon=1e-5):
    n = len(X[0])
    w = [0.0] * n
    b = 0.0
    j_prev = float('inf')
    history = []
    logged_iters = set()
    recent_logs = []

    def log_iteration(iter_idx, w, b, current_cost, cost_diff):
        # 输出当前迭代的参数和代价函数信息
        print(
            f"[{iter_idx:>6}/{num_iters}]: w = {[round(value, 6) for value in w]}, "
            f"b = {b:.6f}, j_wb = {current_cost:.12f}, "
            f"j_prev - j_wb = {cost_diff}"
        )
        logged_iters.add(iter_idx)

    for i in range(num_iters):
        # 1. 计算当前代价函数值
        current_cost = cost_function(X, y, w, b)
        history.append(current_cost)

        cost_diff = j_prev - current_cost if j_prev != float('inf') else float('inf')
        current_log = (i, w[:], b, current_cost, cost_diff)
        recent_logs.append(current_log)
        recent_logs = recent_logs[-2:]

        if i % 10000 == 0:
            log_iteration(*current_log)

        # 2. 判断是否收敛
        if abs(j_prev - current_cost) < epsilon:
            break

        j_prev = current_cost

        # 3. 若未收敛，则使用梯度下降更新参数
        w, b = gradient_descent(X, y, w, b, alpha)

    for final_log in recent_logs:
        if final_log[0] not in logged_iters:
            log_iteration(*final_log)

    return w, b, history
```

```python
# 使用前面定义的函数训练房价预测模型并做预测
X = [
    [60, 2, 5, 10],
    [75, 3, 8, 5],
    [90, 3, 12, 3],
    [110, 4, 15, 1],
    [55, 2, 3, 20],
    [130, 4, 20, 2],
    [85, 3, 6, 12],
    [100, 3, 10, 8],
    [70, 2, 18, 0],
    [120, 4, 25, 6],
]

y = [167.5, 227, 267, 330.5, 134.5, 376, 230, 274, 227, 355.5]

w, b, history = train_model(X, y, alpha=1e-6, num_iters=100000, epsilon=1e-5)

print(f"【训练完成】: w = {[round(value, 6) for value in w]}, b = {b:.6f}, j_wb = {history[-1]:.12f}, epsilon = 1e-5")

# 使用训练后的模型预测新房源的房价
X_test = [
    [60, 2, 5, 10],
    [80, 3, 9, 6],
    [105, 4, 16, 4],
    [65, 2, 4, 18],
    [120, 4, 25, 6],
]
predictions = [sum(w[j] * sample[j] for j in range(len(w))) + b for sample in X_test]

for sample, pred in zip(X_test, predictions):
    print(f"房屋特征 x = {sample}, 【预测房价】 = {pred:.2f} 万元")
```

## 3. 多元线性回归公式

多元线性回归模型可以写成：

$$
\hat{y} = w_1 x_1 + w_2 x_2 + \cdots + w_n x_n + b
$$

在本案例中，房屋有 4 个特征，因此预测房价可以写成：

$$
\hat{y} = w_1 x_1 + w_2 x_2 + w_3 x_3 + w_4 x_4 + b
$$

也可以写成向量形式：

$$
\hat{y} = \vec{w} \cdot \vec{x} + b
$$

如果写到第 $i$ 个房屋样本，则有：

$$
\hat{y}^{(i)} = \vec{w} \cdot \vec{x}^{(i)} + b
$$

其中：

- $\vec{x}^{(i)} = [x_1^{(i)}, x_2^{(i)}, x_3^{(i)}, x_4^{(i)}]$ 表示第 $i$ 个房屋样本的特征向量
- $\vec{w} = [w_1, w_2, w_3, w_4]$ 表示面积、卧室数、楼层、房龄对应的权重
- $b$ 表示偏置，可以理解为基础房价
- $\hat{y}^{(i)}$ 表示模型对第 $i$ 个房屋样本的预测房价

## 4. 向量与向量点积

在多元线性回归中，一个样本通常不再只有一个特征，而是由多个特征组成。对房价预测来说，一个房屋样本可以同时包含面积、卧室数、楼层和房龄。

例如，一个房屋样本可以写成：

$$
\vec{x} = [x_1, x_2, x_3, x_4]
$$

对应地，模型参数也可以写成一个向量：

$$
\vec{w} = [w_1, w_2, w_3, w_4]
$$

所谓“向量点积”，就是把两个向量中对应位置的元素分别相乘，再把结果相加。

在本案例中：

$$
\vec{w} \cdot \vec{x} = w_1x_1 + w_2x_2 + w_3x_3 + w_4x_4
$$

所以，房价预测模型可以写成：

$$
\hat{y} = \vec{w} \cdot \vec{x} + b
$$

本质上只是把逐项展开的式子：

$$
\hat{y} = w_1x_1 + w_2x_2 + w_3x_3 + w_4x_4 + b
$$

压缩成了更简洁的向量写法。

## 5. 代价函数（Cost Function）

在多特征房价预测中，常用均方误差形式的代价函数，衡量预测房价和真实房价之间的整体差距：

$$
J(\vec{w}, b) = \frac{1}{2m} \sum_{i=1}^{m} (\hat{y}^{(i)} - y^{(i)})^2
$$

把预测公式代入后可得：

$$
J(\vec{w}, b) = \frac{1}{2m} \sum_{i=1}^{m} (\vec{w} \cdot \vec{x}^{(i)} + b - y^{(i)})^2
$$

其中 $m$ 表示房屋样本数量，$y^{(i)}$ 表示第 $i$ 个样本的真实房价。

## 6. 更新参数的公式

对每一个权重 $w_j$，梯度为：

$$
\frac{\partial J(\vec{w}, b)}{\partial w_j} = \frac{1}{m} \sum_{i=1}^{m} (\hat{y}^{(i)} - y^{(i)}) x_j^{(i)}
$$

对偏置 $b$，梯度为：

$$
\frac{\partial J(\vec{w}, b)}{\partial b} = \frac{1}{m} \sum_{i=1}^{m} (\hat{y}^{(i)} - y^{(i)})
$$

因此，参数更新公式为：

$$
w_j := w_j - \alpha \frac{\partial J(\vec{w}, b)}{\partial w_j}, \quad j = 1, 2, 3, 4
$$

$$
b := b - \alpha \frac{\partial J(\vec{w}, b)}{\partial b}
$$

其中 $\alpha$ 是学习率。对房价样本这种不同特征量级差异较大的数据，学习率通常要取得更小，或者先对特征做归一化。

## 7. 训练过程示例

下面用这 10 个房屋样本说明一次梯度下降是如何更新参数的。为了便于说明，先设初始参数为：

$$
w_1 = 0, \quad w_2 = 0, \quad w_3 = 0, \quad w_4 = 0, \quad b = 0
$$

学习率取为：

$$
\alpha = 0.000001
$$

此时模型预测公式为：

$$
\hat{y} = 0 \cdot x_1 + 0 \cdot x_2 + 0 \cdot x_3 + 0 \cdot x_4 + 0 = 0
$$

也就是说，当前所有房屋样本的预测房价都为 0。

***第一步：计算代价函数***

共有 $m = 10$ 个房屋样本，因此：

$$
J(\vec{w}, b) = \frac{1}{2 \times 10} \sum_{i=1}^{10} (\hat{y}^{(i)} - y^{(i)})^2
$$

由于当前所有 $\hat{y}^{(i)} = 0$，所以：

$$
J(0,0,0,0,0) = \frac{1}{20} (167.5^2 + 227^2 + 267^2 + 330.5^2 + 134.5^2 + 376^2 + 230^2 + 274^2 + 227^2 + 355.5^2)
$$

$$
= \frac{725456}{20} = 36272.8
$$

***第二步：计算 $w_1$ 的梯度***

$$
\frac{\partial J}{\partial w_1} = \frac{1}{10} \sum_{i=1}^{10} (\hat{y}^{(i)} - y^{(i)}) x_1^{(i)}
$$

由于 $\hat{y}^{(i)} = 0$，代入面积数据可得：

$$
\frac{\partial J}{\partial w_1} = \frac{-249237.5}{10} = -24923.75
$$

同理可得其他特征和偏置的梯度：

$$
\frac{\partial J}{\partial w_2} = -830, \quad
\frac{\partial J}{\partial w_3} = -3583.2, \quad
\frac{\partial J}{\partial w_4} = -1446.85, \quad
\frac{\partial J}{\partial b} = -258.9
$$

***第三步：更新参数***

$$
w_1 := 0 - 0.000001 \times (-24923.75) = 0.02492375
$$

$$
w_2 := 0 - 0.000001 \times (-830) = 0.00083
$$

$$
w_3 := 0 - 0.000001 \times (-3583.2) = 0.0035832
$$

$$
w_4 := 0 - 0.000001 \times (-1446.85) = 0.00144685
$$

$$
b := 0 - 0.000001 \times (-258.9) = 0.0002589
$$

所以，一次更新后参数从：

$$
(w_1, w_2, w_3, w_4, b) = (0, 0, 0, 0, 0)
$$

变成：

$$
(w_1, w_2, w_3, w_4, b) = (0.02492375, 0.00083, 0.0035832, 0.00144685, 0.0002589)
$$

## 8. 如何判断收敛

常见的一种方式，是比较这一轮和上一轮的代价函数变化是否已经足够小：

$$
\left| J_{\text{prev}} - J(\vec{w}, b) \right| < \epsilon
$$

其中：

- $J_{\text{prev}}$ 表示上一轮的代价函数值
- $J(\vec{w}, b)$ 表示当前这一轮的代价函数值
- $\epsilon$ 表示收敛阈值，是一个很小的正数

如果满足这个条件，就可以认为房价预测模型已经基本收敛。

## 9. 训练过程总结

多特征房价预测的训练目标，就是不断调整参数向量 $\vec{w}$ 和偏置 $b$，让代价函数 $J(\vec{w}, b)$ 尽可能小，也就是让预测房价尽量接近真实房价。

| 符号 | 释义 |
| --- | --- |
| $$\{(\vec{x}^{(1)}, y^{(1)}), (\vec{x}^{(2)}, y^{(2)}), \ldots, (\vec{x}^{(m)}, y^{(m)})\}$$ | 房屋训练数据 |
| $\vec{x}$ | 房屋特征向量，包含面积、卧室数、楼层、房龄 |
| $y$ | 真实房价 |
| $\hat{y}$ | 预测房价 |
| $\vec{w}$ | 权重向量 |
| $b$ | 偏置，也可以理解为基础房价 |
| $\alpha$ | 学习率 |
| $\text{num\_iters}$ | 最大迭代次数 |
| $\epsilon$ | 收敛阈值 |
| $J(\vec{w}, b)$ | 代价函数 |
| $J_{\text{prev}}$ | 上一轮的代价函数值 |

***初始化：***

$$
\vec{w} = [0, 0, 0, 0], \quad b = 0, \quad J_{\text{prev}} = +\infty
$$

***for (i=0, i<num_iters, i++):***

***{***

$$
\begin{aligned}
&\text{1. 计算当前代价函数：其中 } \hat{y}^{(i)} = \vec{w} \cdot \vec{x}^{(i)} + b\\
&J(\vec{w}, b) = \frac{1}{2m} \sum_{i=1}^{m} (\hat{y}^{(i)} - y^{(i)})^2 = \frac{1}{2m} \sum_{i=1}^{m} (\vec{w} \cdot \vec{x}^{(i)} + b - y^{(i)})^2\\[6pt]
&\text{2. 判断是否收敛：}\\
&\text{若 } \left| J_{\text{prev}} - J(\vec{w}, b) \right| < \epsilon \text{，则停止训练；否则令 } J_{\text{prev}} = J(\vec{w}, b)\\[6pt]
&\text{3. 若未收敛，则更新参数：}\\
&w_j = w_j - \alpha \frac{\partial J(\vec{w}, b)}{\partial w_j} = w_j - \alpha \frac{1}{m} \sum_{i=1}^{m} (\hat{y}^{(i)} - y^{(i)}) x_j^{(i)}, \quad j = 1, 2, 3, 4\\[6pt]
&b = b - \alpha \frac{\partial J(\vec{w}, b)}{\partial b} = b - \alpha \frac{1}{m} \sum_{i=1}^{m} (\hat{y}^{(i)} - y^{(i)})\\
\end{aligned}
$$

***}***

最终输出训练后的参数向量 $\vec{w}$ 和偏置 $b$，然后用它们预测新房源的房价。
